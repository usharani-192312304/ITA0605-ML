```python
# FIND-S Algorithm

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High',    'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High',    'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High',    'Strong', 'Cool', 'Change', 'Yes']
]

# Initialize most specific hypothesis
h = ['0'] * 6

for row in data:
    if row[-1] == 'Yes':
        for i in range(6):
            if h[i] == '0':
                h[i] = row[i]
            elif h[i] != row[i]:
                h[i] = '?'

print("Most Specific Hypothesis:", h)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

# Dataset
data = [
    ['Sunny', 'Hot', 'High', 'Weak', 'No'],
    ['Sunny', 'Hot', 'High', 'Strong', 'No'],
    ['Overcast', 'Hot', 'High', 'Weak', 'Yes'],
    ['Rain', 'Mild', 'High', 'Weak', 'Yes'],
    ['Rain', 'Cool', 'Normal', 'Weak', 'Yes'],
    ['Rain', 'Cool', 'Normal', 'Strong', 'No'],
    ['Overcast', 'Cool', 'Normal', 'Strong', 'Yes'],
    ['Sunny', 'Mild', 'High', 'Weak', 'No'],
    ['Sunny', 'Cool', 'Normal', 'Weak', 'Yes'],
    ['Rain', 'Mild', 'Normal', 'Weak', 'Yes']
]

# Convert text data to numbers
encoders = []
X = []

for i in range(4):
    le = LabelEncoder()
    X.append(le.fit_transform([row[i] for row in data]))
    encoders.append(le)

X = list(zip(*X))
y = LabelEncoder().fit_transform([row[4] for row in data])

# Build ID3 tree
model = DecisionTreeClassifier(criterion='entropy', random_state=0)
model.fit(X, y)

# New sample: Sunny, Cool, High, Strong
sample = [[
    encoders[0].transform(['Sunny'])[0],
    encoders[1].transform(['Cool'])[0],
    encoders[2].transform(['High'])[0],
    encoders[3].transform(['Strong'])[0]
]]

prediction = model.predict(sample)

print("Predicted Class:", "Yes" if prediction[0] == 1 else "No")
```

```python
# Candidate Elimination Algorithm

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

S = ['0'] * 6
G = [['?'] * 6]

for row in data:
    x, y = row[:-1], row[-1]

    if y == 'Yes':
        G = [g for g in G if all(
            g[i] == '?' or g[i] == x[i] for i in range(6))]

        for i in range(6):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

    else:
        new_G = []
        for g in G:
            for i in range(6):
                if g[i] == '?' and S[i] != x[i]:
                    h = g.copy()
                    h[i] = S[i]
                    new_G.append(h)
        G = new_G

print("Specific Boundary (S):", S)
print("General Boundary (G):", G)
```
